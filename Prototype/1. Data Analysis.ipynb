```python
# Importing libraries for data analysis and loading dataset
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading dataset
file_url = "alzheimers_disease_data.csv"
dataframe = pd.read_csv(file_url)

# Checking the first 5 rows of data in tabular form
display(dataframe.head())
```

```python
# Checking the rows and columns of the dataset
print("Shape of the DataFrame:")
print(dataframe.shape)
print("\nColumns in the DataFrame:")
print(dataframe.columns)
```

```python
# Display summary statistics of numerical columns
print("\nSummary Statistics:")
display(dataframe.describe())
```

```python
# Check for any missing values
print("\nMissing Values:\n")
print(dataframe.isnull().sum())
```

```python
# Check for any duplicate rows
print(f"\nDuplicate Rows: {dataframe.duplicated().sum()}")
```

```python
# Checking datatypes of all columns
print("\nData types of the columns:")
print(dataframe.dtypes)
```

```python
# Class distribution of dataset with Diagnosis
plt.figure(figsize=(8, 5))
sns.countplot(x='Diagnosis', data=dataframe)
plt.title('Class Distribution')
plt.show()
```

```python
# Selecting only numerical columns
numerical_columns = dataframe.select_dtypes(include=['int64', 'float64'])

# Calculating correlations
correlations = numerical_columns.corr()

# Creating a basic heatmap to visualise the correlations
correlation_matrix = numerical_columns.corr()
plt.figure(figsize=(20,16))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Pairwise Correlation Heatmap')
plt.show()
```
